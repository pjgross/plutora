```python
import plutora as p
tecrs = p.Get_TECRs("2024-01-01", "2024-12-31")
tecrs
```

```python
import pandas as pd
import plotly.express as px

# Ensure your Start Date column is in datetime format
tecrs['Start Date'] = pd.to_datetime(tecrs['Start Date'], errors='coerce')

# Extract month-year for grouping
tecrs['Month'] = tecrs['Start Date'].dt.to_period('M').astype(str)  # e.g. '2024-01'

# Count how many releases per month
monthly_counts = tecrs.groupby('Month').size().reset_index(name='Release Count')

# Create a bar chart
fig = px.bar(
    monthly_counts,
    x='Month',
    y='Release Count',
    title='Number of Releases per Month',
    labels={'Month': 'Month', 'Release Count': 'Number of Releases'},
    text='Release Count'
)

# Sort x-axis chronologically (optional)
fig.update_layout(
    xaxis={'categoryorder': 'category ascending'},
    template='plotly_white'
)

fig.show()
```

```python
import pandas as pd
import plotly.express as px

# Ensure Start Date is datetime
tecrs['Start Date'] = pd.to_datetime(tecrs['Start Date'], errors='coerce')

# Create a month-period column for sorting
tecrs['MonthPeriod'] = tecrs['Start Date'].dt.to_period('M')

# Create a readable month label
tecrs['Month'] = tecrs['MonthPeriod'].dt.strftime('%b %Y')

# Group by MonthPeriod and TECR Status (to preserve order)
monthly_status_counts = (
    tecrs.groupby(['MonthPeriod', 'Month', 'TECR Status'])
    .size()
    .reset_index(name='Count')
    .sort_values('MonthPeriod')  # ensures chronological order
)

# Plot
fig = px.bar(
    monthly_status_counts,
    x='Month',
    y='Count',
    color='TECR Status',
    title='TECRs per month by Status',
    labels={'Month': 'Month', 'Count': 'Number of TECRs', 'TECR Status': 'Status'},
    text='Count'
)

# Format layout
fig.update_layout(
    barmode='stack',
    xaxis={'categoryorder': 'array', 'categoryarray': monthly_status_counts['Month'].unique()},
    template='plotly_white'
)

fig
```

```python
import pandas as pd
import requests

# Example: you already have a dataframe
# df = pd.read_csv("input.csv")

# Function to call your API for one row
def call_api(row):
    token = p.getAccessToken()
    url = f"https://ukapi.plutora.com/TECRs/{row.id}/additionalInformation"
    headers = {
        'accept': 'application/json',
        'Plutora-Info': 'script.name=swagger',
        'Authorization': f'bearer {token}'
    }

    try:
        response = requests.get(url, headers=headers, timeout=10)
        items = response.json()
        result = {}
        for i, item in enumerate(items):
            print(f"{item}")
            name = item['name']
            data_type = item.get("dataType")

            # Pick the correct value depending on dataType
            if data_type == "FreeText":
                if item.get("text") == None:
                    result[name] = ""
                else:
                    result[name] = item.get("text")
            elif data_type == "ListField":
                if item.get("listItem") == None:
                    result[name] = ""
                else:
                    result[name] = item.get("listItem", {}).get("value")
            else:
                result[name] = ""

        return result
        # return a dict of the values you want to add


    except requests.RequestException as e:
        # handle network/API errors gracefully
        return {
            "api_status": None,
            "api_score": None,
            "api_message": f"Error: {e}"
        }

# Apply the function to each row and expand the dict into columns
results = tecrs.apply(call_api, axis=1, result_type='expand')

# Merge results back into the original dataframe
df = pd.concat([tecrs, results], axis=1)
```
